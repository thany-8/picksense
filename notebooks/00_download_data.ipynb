# Download Data

```python
from huggingface_hub import hf_hub_download
from pathlib import Path
import zipfile

# Find project root by looking for a marker file
def find_project_root(marker="pyproject.toml"):
    current = Path.cwd()
    for parent in [current, *current.parents]:
        if (parent / marker).exists():
            return parent
    return current  # fallback

PROJECT_ROOT = find_project_root()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw data:", RAW_DIR)
```

```python
REPO = "zeredata/bin-picking"

train_zip = hf_hub_download(
    repo_id=REPO,
    filename="bin_picking_train_bop.zip",
    repo_type="dataset"
)

print(train_zip)
```

```python
with zipfile.ZipFile(train_zip, "r") as z:
    z.extractall(RAW_DIR)

print("Done!")
```

```python
train_dir = RAW_DIR / "data" / "train"

print(train_dir)
print("Exists:", train_dir.exists())

if train_dir.exists():
    print("Scenes:", len(list(train_dir.iterdir())))
```
